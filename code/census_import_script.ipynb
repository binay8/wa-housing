# Census Data Import

```python

# Using the following code we will be importing Median income for all counties in Washington state

API_KEY = "08da6ee1bacb25b887038df0d6c2bc19e0115207"

all_years = []

for year in range(2016, 2025):
    url = f"https://api.census.gov/data/{year}/acs/acs5"

    params = {
        "get": "NAME,B19013_001E",
        "for": "county:*",
        "in": "state:53",
        "key": API_KEY
    }

    response = requests.get(url, params=params, timeout=30)
    response.raise_for_status()

    data = response.json()

    # First row contains column names
    df = pd.DataFrame(data[1:], columns=data[0])

    df["Year"] = year

    all_years.append(df)

# Combine every year into a single DataFrame
income = pd.concat(all_years, ignore_index=True)

# Rename columns
income = income.rename(columns={
    "NAME": "County",
    "B19013_001E": "MedianIncome"
})

# Convert income to numeric
income["MedianIncome"] = pd.to_numeric(
    income["MedianIncome"], errors="coerce"
)

# Preserve geographic identifiers
income["CountyFIPS"] = income["state"] + income["county"]

# Check results
print(income.shape)
print(income.head())

# Export
income.to_csv("C:/Users/binay/Documents/data5100/projects/WAHousing/data/WA_Median_Income_2015_2024.csv", index=False)
```
