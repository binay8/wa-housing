Notes on housing value dataset:
* The dataset currently includes all states. We need to filter and get WA state data.
* County information is stored in "RegionName" column.
* Additionally, Monthly home values (ZHVI) are stored in separate date columns.

```python
# Identify columns containing monthly dates
date_cols = [
    col for col in housing_value.columns
    if col[:4].isdigit()
]

# Convert to tidy format
housing_tidy = housing_value.melt(
    id_vars=["RegionID", "RegionName", "State",
             "StateCodeFIPS", "MunicipalCodeFIPS"],
    value_vars=date_cols,
    var_name="Date",
    value_name="HomeValue"
)

housing_tidy["Date"] = pd.to_datetime(housing_tidy["Date"])
```

```python
housing_tidy
```

```python
wa_housing = housing_tidy[
    (housing_tidy["State"] == "WA") &
    (housing_tidy["Date"].dt.year.between(2015, 2024))
].copy()
wa_housing
```

```python
print("Shape:", wa_housing.shape)
print("Total County (RegionName):",wa_housing["RegionName"].nunique())
print("Minimum Date (Earliest):",wa_housing["Date"].min())
print("Maximum Date (Latest):",wa_housing["Date"].max())
print("Total Missing Data:\n",wa_housing.isna().sum())
```

```python
# Calcualted column "CountyFIPS" based on StateCodeFIPS and Municipal CodeFIPS
wa_housing["CountyFIPS"] = wa_housing["StateCodeFIPS"].astype(str).str.zfill(2) + wa_housing["MunicipalCodeFIPS"].astype(str).str.zfill(3)
```

```python
wa_housing.columns
```

* So far we have converted the dataset to tidy format
* Dataset has been filtered to include: Washington state only
* Dataset has been filtered to include: from years 2015 to 2024
* It seems like there is one row with missing HomeValue

```python
wa_housing.info()
```

```python
# Investigate the missing homevalue 
wa_housing[wa_housing["HomeValue"].isna()]
```

* We are missing Home value data for the entire year in 2015 for Columbia County and 1 month in 2016.
* For now we will leave this nan as we have to group by county, year to calculate yearly average.
* Since Home value data for the entire year of 2015 is missing for Columbi acounty, we will have one row with no average house value.

```python
# Extract year
wa_housing["Year"] = wa_housing["Date"].dt.year

# Aggregate monthly ZHVI into annual values
housing_annual = wa_housing.groupby(
        ["CountyFIPS", "RegionName", "Year"],
        as_index=False
    ).agg(
        AvgHomeValue=("HomeValue", "mean"),
        MonthsAvailable=("HomeValue", "count")
    )


housing_annual.head()
```

```python
housing_annual.info()
```

```python
housing_clean = housing_annual.rename(columns={"RegionName": "county",
                                               "Year": "year", 
                                               "CountyFIPS": "county_fips", 
                                               "AvgHomeValue": "average_zhvi"}).copy()
housing_clean.info()
```

```python
housing_clean = housing_clean[['county_fips', 'county', 'year', 'average_zhvi']]
housing_clean.head()
```

```python
housing_clean.to_csv(
    DATA_DIR / 'housing_clean.csv',
    encoding='utf-8-sig',
    index=False
)
```

## Import Unemployment Rate (BLS)

```python
IN_PATH = DATA_DIR / "WA_Unemployment_2015_2024.csv"


OUT_PATH = DATA_DIR / "cleaned_unemployment_rate.csv"
 
SEATTLE_METRO = ["King", "Pierce", "Snohomish"]
EASTERN_WA = ["Adams", "Ferry", "Lincoln", "Pend Oreille", "Spokane", "Stevens", "Whitman"]
```

```python
# Load data set
df = pd.read_csv(IN_PATH, dtype={"CountyFIPS": str})
```

```python
df.head()
```

```python
df.info()
```

* The dataset contains 390 observations with no missing values.
* County FIPS is stored as a string to preserve its identifier format.
* County names include the state name and will be standardized.
* Column names will be standardized for merging.

```python
# Clean data set
df = df.rename(columns={
    "CountyFIPS": "county_fips",
    "County": "county_full",
    "Year": "year",
    "UnemploymentRate": "unemployment_rate",
})
df["county_fips"] = df["county_fips"].str.strip().str.zfill(5)
df["year"] = pd.to_numeric(df["year"], errors="coerce").astype("Int64")
df["unemployment_rate"] = pd.to_numeric(df["unemployment_rate"], errors="coerce")
 
# "Adams County, Washington" -> "Adams"
df["county"] = (
    df["county_full"].str.replace(", Washington", "", regex=False)
                     .str.replace(" County", "", regex=False)
                     .str.strip()
)
df = df.drop(columns="county_full")
 
n_raw = len(df)
df = df.dropna(subset=["county_fips", "year", "unemployment_rate"]).drop_duplicates(["county_fips", "year"])
print(f"Rows: {n_raw} raw -> {len(df)} after cleaning")
```

```python
unemployment_clean = df[['county_fips', 'year', 'unemployment_rate']]
unemployment_clean.head()
```

```python
unemployment_clean.info()
```

```python
unemployment_clean.to_csv(
    OUT_PATH,
    encoding='utf-8-sig',
    index=False
)
```

## Import Interest Rate

```python
IN_PATH = DATA_DIR / "interest_rate.csv"
OUT_PATH = DATA_DIR / "cleaned_annual_interest_rate.csv"
START_YEAR, END_YEAR = 2015, 2024
```

```python
# Load data set
df = pd.read_csv(IN_PATH)
```

```python
# Clean data set
df = df.rename(columns={"observation_date": "date", "MORTGAGE30US": "rate"})
df["date"] = pd.to_datetime(df["date"], errors="coerce")
# FRED sometimes writes "." for missing values; coerce turns those into NaN
df["rate"] = pd.to_numeric(df["rate"], errors="coerce")
 
n_raw = len(df)
df = (
    df.dropna(subset=["date", "rate"])   # drop unparseable dates / missing rates
      .drop_duplicates(subset="date")    # one observation per week
      .sort_values("date")
      .reset_index(drop=True)
)
print(f"Rows: {n_raw} raw -> {len(df)} after cleaning")
# Sanity check: rates should be plausible percentages
assert df["rate"].between(0, 25).all(), "Rate outside 0-25% - check the data"
```

```python
df.info()
```

```python
# Keep only the study window: 2015 - 2024
df["year"] = df["date"].dt.year
df = df[df["year"].between(START_YEAR, END_YEAR)]
```

```python
df.head()
```

```python
# Aggregate weekly -> yearly
yearly = (
    df.groupby("year")["rate"]
      .agg(
          mortgage_rate_mean="mean",      # primary variable for the model
          mortgage_rate_median="median",
          mortgage_rate_min="min",
          mortgage_rate_max="max",
          n_weeks="count",
      )
      .reset_index()
)

# Year-end rate (last weekly observation of each year), as an alternative measure
yearly["mortgage_rate_yearend"] = df.groupby("year")["rate"].last().values
 
# Change in the annual average rate vs. the prior year (percentage points)
yearly["mortgage_rate_change"] = yearly["mortgage_rate_mean"].diff()
yearly = yearly.round(3)
```

```python
# Validate
expected_years = set(range(START_YEAR, END_YEAR + 1))
assert set(yearly["year"]) == expected_years, "Missing years in output"
short = yearly[yearly["n_weeks"] < 50]
if not short.empty:
    print("WARNING: years with fewer than 50 weekly observations:\n", short[["year", "n_weeks"]])
```

```python
# Save new file
yearly.to_csv(OUT_PATH, index=False)
print(yearly.to_string(index=False))
print(f"\nSaved to {OUT_PATH}")
```
